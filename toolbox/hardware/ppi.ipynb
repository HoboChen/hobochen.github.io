```python
import math

def calculate_dimensions(x_pixels, y_pixels, diagonal_inches):
    diagonal_pixels = math.sqrt(x_pixels**2 + y_pixels**2)
    ppi = diagonal_pixels / diagonal_inches
    x_length = (x_pixels * diagonal_inches) / diagonal_pixels
    y_length = (y_pixels * diagonal_inches) / diagonal_pixels
    x_length *= 25.4
    y_length *= 25.4
    m_pixels = x_pixels * y_pixels / 1e6
    return (x_length, y_length, ppi, m_pixels)
```

```python
monitors = [
    ("Dell U2725QE",         "27\" 4K",    3840, 2160, 27),
    ("ViewSonic VP2488-4K",  "23.8\" 4K",  3840, 2160, 23.8),
    ("iMac 24 M4",           "23.5\" 4.5K", 4480, 2520, 23.5),
    ("BenQ MA282",           "28.2\" 3:2", 3840, 2560, 28.2),
    ("Apple Studio Display", "27\" 5K",    5120, 2880, 27),
    ("Dell P3225QE",         "31.5\" 4K",  3840, 2160, 31.5),
    ("LG 32U990A",           "31.5\" 6K",  6144, 3456, 31.5),
    ("Sony A90K 42",         "42\" 4K",    3840, 2160, 42),
]

print(f"{'Monitor':<24} {'Spec':<14} {'Size (mm)':<20} {'PPI':>6} {'MP':>6}")
print("-" * 74)
for name, spec, x, y, diag in monitors:
    w, h, ppi, mp = calculate_dimensions(x, y, diag)
    print(f"{name:<24} {spec:<14} {w:.0f} x {h:.0f} mm     {ppi:>6.0f} {mp:>6.1f}")
```
